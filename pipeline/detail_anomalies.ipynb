<h1 style = "text-align: center; color: green">Détail des anomalies ressorties du catalogue</h1>

```python
import pandas as pd
```

```python
df = pd.read_csv("test_manifests.csv")
df.head(2)
```

```python
df = df[df["anomalies"].notna()]
df["anomalies"].value_counts()
```

<h2 style= "text-align:center; color: green">Doublon sha256</h2>

```python
df_sha256 = df[df["anomalies"] == "Doublon SHA256"]

df_sha256 = df_sha256.groupby(["livraison", "sha256"])["fichier_origine"].agg(list)
df_sha256 = df_sha256.reset_index()

df_sha256["fichier_origine_2"] = df_sha256["fichier_origine"]

df_sha256["fichier_origine"] = df_sha256["fichier_origine"].str[0]
df_sha256["fichier_origine_2"] = df_sha256["fichier_origine_2"].str[1]

df_sha256
```

<h2 style= "text-align:center; color: green">Même individu, même côté, sha différent</h2>

```python
df_cote = df[df["anomalies"] == "même indivu, même côté, sha256 différents"]


df_cote = df_cote.groupby(["livraison", "individu", "cote"])[
    ["fichier_origine", "sha256"]
].agg(list)
df_cote = df_cote.reset_index()

df_cote["fichier_origine_2"] = df_cote["fichier_origine"].copy()
df_cote["sha256_2"] = df_cote["sha256"].copy()

df_cote["fichier_origine"] = df_cote["fichier_origine"].str[0]
df_cote["fichier_origine_2"] = df_cote["fichier_origine_2"].str[1]
df_cote["sha256"] = df_cote["sha256"].str[0]
df_cote["sha256_2"] = df_cote["sha256_2"].str[1]

df_cote
```
